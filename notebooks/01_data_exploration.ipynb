# Data Exploration - HAM10000


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
# Load metadata
metadata_path = '../data/raw/HAM10000_metadata.csv'
if os.path.exists(metadata_path):
    df = pd.read_csv(metadata_path)
    print(df.head())
else:
    print('Please download dataset first.')
```

### Basic Statistics


```python
if 'df' in locals():
    print(f'Total Images: {len(df)}')
    print(f'Number of Classes: {df["dx"].nunique()}')
    print('\nClass Distribution:')
    print(df['dx'].value_counts())
    
    plt.figure(figsize=(10,6))
    sns.countplot(data=df, x='dx')
    plt.title('Class Distribution in HAM10000')
    plt.show()
```

### Missing Values Analysis


```python
if 'df' in locals():
    print('Missing values per column:')
    print(df.isnull().sum())
    
    # Age distribution
    plt.figure(figsize=(10,6))
    sns.histplot(df['age'].dropna(), bins=20, kde=True)
    plt.title('Age Distribution')
    plt.show()
```
